**Euler Method and  Euler Error Analysis**

The Euler Method is a first-order numerical method for approximating the solution of an initial-value problem (IVP):

dy/dx =f(x,y)     with IVP   y(x0)=y0


It uses the slope at the current point to estimate the solution at the next point.



**What Euler Method actually does?**

Euler Method is used to find an approximate solution of a differential equation when solving it exactly is difficult or when we want a numerical solution.

**Formula:**

yn+1 = yn+hf(xn+yn)

xn+1 = xn + h

where

1. h = step size

2. xn = current value of x

3. yn = current numerical approximation

4. f(xn,yn)  = slope at the current point

**Algorithm:**

**Step 1**

Given the differential equation

dy/dx =f(x,y)

and initial condition

y(x0)=y0

**Step 2**

Choose the step size h.

**Step 3**

Calculate the slope:

fn=f(xn,yn)

**Step 4**

Calculate the next approximation:

yn+1=yn +h* fn

	​
**Step 5**

Update x:

xn+1 = xn + h.

**Step 6**

Repeat Steps 3–5 until the required endpoint or number of steps is reached.

**Step 7**

Store and display the numerical solution.


```python
# QUESTION NO 01:
# Use the Euler Method to approximate the solution of the initial-value problem
# dy/dx=x+y,y(0)=1
# on the interval 0≤x≤1 using a step size h=0.1. Display the numerical approximations and plot the computed solution.
import numpy as np
import matplotlib.pyplot as plt
import math
def f(x,y):
  return x+y
x=0
y=1
h=0.1
tol=1e-6
n=3
x_values=[]
y_values=[]
for i in range(n):
  x_values.append(x)
  y_values.append(y)
  slope=f(x,y)
  y_new=y+h*slope
  x_new=x+h
  x=x_new
  y=y_new
  print("iterations:  ", i+1)
  print("x", x)
  print("y   ", y)
  print("slope  ", slope)
plt.plot(
    x_values,
    y_values,
    marker="o",
    label="Euler Approximation"
)
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()
plt.show()



```

Therefore:

smaller h→usually better accuracy
	​

**QUESTION NO 04:**

Use the Euler Method to solve the differential equation


y' = x+y

with the initial condition (y(0)=1). Use a step size of (h=0.1) and calculate the solution up to (x=0.5). Compare the numerical Euler solution with the exact solution and plot both results on the same graph.

```python
import numpy as np
import matplotlib.pyplot as plt

# Differential equation
def f(x, y):
    return x + y


# Exact solution
def exact_solution(x):
    return 2*np.exp(x) - x - 1


# Initial values
x0 = 0
y0 = 1

# Step size
h = 0.1

# Final x
x_final = 0.5

# Number of iterations
n = int((x_final - x0) / h)

print("Number of iterations =", n)


# Store Euler values
x_values = [x0]
y_values = [y0]

# Current values
x = x0
y = y0


# Euler Method
for i in range(n):

    # Calculate slope
    slope = f(x, y)

    # Euler formula
    y_new = y + h * slope

    # Next x
    x_new = x + h

    # Update
    x = x_new
    y = y_new

    # Store values
    x_values.append(x)
    y_values.append(y)

    print("Iteration", i + 1,
          ": x =", x,
          ", y =", y)


# Smooth x-values for exact solution
X_exact = np.linspace(x0, x_final, 500)

# Exact y-values
Y_exact = exact_solution(X_exact)


# Plot exact solution
plt.figure(figsize=(8, 6))

plt.plot(
    X_exact,
    Y_exact,
    linewidth=2,
    label="Exact Solution"
)


# Plot Euler approximation
plt.plot(
    x_values,
    y_values,
    marker="o",
    linewidth=2,
    label="Euler Approximation"
)


plt.xlabel("x")
plt.ylabel("y")
plt.title("Euler Method vs Exact Solution")

plt.grid(True)
plt.legend()

plt.show()
```

**QUESTION NO 05:**

Use the Euler Method to solve the differential equation (y'=x+y) with the initial condition (y(0)=1). Use a step size of (h=0.1) and calculate the solution up to (x=0.5). Compare the Euler approximation with the exact solution (y(x)=2e^x-x-1), and plot both solutions on the same graph.

```python
#Kyunke hum fixed final x=0.5 tak iterations kar rahe hain. tol ki zaroorat tab hogi jab hum error/convergence ke basis par stopping condition lagayenge.
import numpy as np
import matplotlib.pyplot as plt
def f(x,y):
  return x+y
def exact_corresponding_solution(x):
  return 2*np.exp(x)-x-1
x0=0
y0=1
h=0.1
x_final=0.5
n = int((x_final - x0) / h)

print("Number of iterations =", n)
x_values=[]
y_values=[]
x=x0
y=y0
for i in range(n):

  slope=f(x,y)
  y_new=y+h*slope
  x_new=x+h
  x=x_new
  y=y_new
  x_values.append(x)
  y_values.append(y)
  print("Iteration", i + 1,
          ": x =", x,
          ", y =", y)
# Smooth x-values for exact solution
X_exact = np.linspace(x0, x_final, 500)

# Exact y-values
Y_exact = exact_corresponding_solution(X_exact)
# Plot exact solution
plt.figure(figsize=(8, 6))

plt.plot(
    X_exact,
    Y_exact,
    linewidth=2,
    label="Exact Solution Graph",
    color='purple'
)


# Plot Euler approximation
plt.plot(
    x_values,
    y_values,
    marker="o",
    linewidth=2,
    label="Euler Approximation",
    color='red'
)


plt.xlabel("x")
plt.ylabel("y")
plt.title("Euler Method vs Exact Solution")

plt.grid(True)
plt.legend()

plt.show()

```

Calculate the absolute error of the Euler Method at each computed (x)-value by comparing the Euler approximation with the exact solution. Display the (x)-value, Euler approximation, exact value, and corresponding error.

**Note:**

If the exact solution y(x) is known, calculate the error at each point.

Absolute Error :

E abs = | y exact − y Euler |

also we can calculate percentage error and realtive error by using:

Relative Error= Absolute error/ |y exact|
	​

Percentage Error=Relative Error×100


```python
print("\nEuler Error Analysis:")

for i in range(len(x_values)):

    exact_y = exact_corresponding_solution(x_values[i])

    # Absolute Error
    absolute_error = abs(exact_y - y_values[i])

    # Relative Error
    if exact_y != 0:
        relative_error = absolute_error / abs(exact_y)
        percentage_error = relative_error * 100
    else:
        relative_error = 0
        percentage_error = 0

    print(
        "x =", x_values[i],
        "Euler =", y_values[i],
        "Exact =", exact_y,
        "Absolute Error =", absolute_error,
        "Relative Error =", relative_error,
        "Percentage Error =", percentage_error, "%"
    )
```
