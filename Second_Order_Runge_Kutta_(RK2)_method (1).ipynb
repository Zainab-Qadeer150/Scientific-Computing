**Second-Order Runge-Kutta (RK2) method:**

This numerical method used to obtain an approximate solution of a first-order ordinary differential equation (ODE):
       dy/dx =f(x,y)

Rk2 takes a basic euler method and evaluates the slope twice in each step and uses these two slopes to get a more accurate approximation of the next value of y.

A common RK2 version is called the **Improved Euler Method (Heun's Method).**

**Importance:**

*   It provides better approximation to solution curve.
*   It is a good intermediate step between the Euler method and RK4.

*   It uses two slope evaluations instead of one.
*  It is generally more accurate than Euler's method.

**Algorithm**

Given:
dy/dx = f(x,y)
x0, y0
step size h

For each iteration:

1. k1 = f(xn, yn)

2. y_predict = yn + h*k1

3. k2 = f(xn+h, y_predict)

4. yn+1 = yn + (h/2)*(k1+k2)

5. xn+1 = xn+h

6. Repeat

7. if x final is given(**tells us where to stop**),we calculate the number of iterations from the step size h and stop when xfinal is reached. Alternatively, we can use a convergence criterion, when xfinal are not given and stop whenever the numerical solution has sufficiently converged means iterations are producing negligible difference.

```python
# Solve f(x,y)=x+y by RK2 method and show the graph:
import numpy as np
import matplotlib.pyplot as plt
def f(x,y):
  return x+y
x0=0
y0=1
h=0.1
x_final=0.5
x_values=[x0]#its stores the RK2 values
y_values=[y0]
x=x0
y=y0
n=int((x_final-x0)/h)
for i in range(n):
  k1=f(x,y)
  y_predict =y+h*k1
  k2=f(x+h,y_predict)
  y_average = y+h/2*(k1+k2)
  x_new = x+h
  x=x_new
  y=y_average
  x_values.append(x)
  y_values.append(y)

print("iteration: ",i+1,
      "\nx_values: ",x,
      "\ny_values: ",y,
      "\nk1 values: ", k1,
      "\nk2 values: ", k2,
      "\ny_average values: ", y_average)
plt.figure(figsize=(8, 6))
plt.plot(x_values, y_values, marker='o', linestyle='-', color='pink',markerfacecolor='black',markeredgecolor='black')
plt.xlabel('x')
plt.ylabel('y')
plt.title('RK2 Method(x+y)')
plt.grid(True)
plt.show()
```

**QUESTION:**

"Write a Python program to solve a first order ordinary differential equation using the Second Order Runge-Kutta (RK2) or Improved Euler Method with a user defined tolerance. The user will enter the ODE, the initial values (x0) and (y0), the step size (h) and the tolerance. The program will calculate (k1), the predicted value of (y), (k2) and the new RK2 approximation and continue the iterations until the change between two consecutive numerical solutions is lesser then the **tolerance.**

 Plot the RK2 approximation by storing the numerical $x$- and $y$-values using Python and Matplotlib, clearly showing the numerical solution points, and label the axs and graph title."

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x, y = sp.symbols('x y')
function = input("Please enter your ODE function dy/dx = ")
Func = sp.sympify(function)
f = sp.lambdify((x, y), Func, modules="numpy")
x0 = float(input("\nEnter x0 value = "))
y0 = float(input("Enter y0 value = "))
h = float(input("Enter h value = "))
tol = float(input("Enter tolerance = "))
x_values = [x0]
y_values = [y0]
x = x0
y = y0
max_iter = 1000
print("\nRK2 Method:")
print("-" * 30)

for i in range(max_iter):
    k1 = f(x, y)
    y_predict = y + h * k1
    k2 = f(x + h, y_predict)
    y_new = y + (h / 2) * (k1 + k2)
    x_new = x + h
    change = abs(y_new - y)
    x_values.append(x_new)
    y_values.append(y_new)
    print(
        "Iteration =", i + 1,
        ": x =", x_new,
        ", y =", y_new,
        ", k1 =", k1,
        ", k2 =", k2,
        ", change =", change
    )

    if change < tol:

        print("\nConvergence achieved!")
        print("Final x =", x_new)
        print("Final y =", y_new)
        print("Tolerance =", tol)
        print("Iterations =", i + 1)

        x = x_new
        y = y_new

        break
    x = x_new
    y = y_new # we are updating the values for next iterations

else:

    print("\nMaximum number of iterations reached.")
    print("Final x =", x)
    print("Final y =", y)
plt.figure(figsize=(8, 6))

plt.plot(
    x_values,
    y_values,
    color="red",
    marker="o",
    markeredgecolor="green",
    markerfacecolor="black",
    linestyle="-",
    markersize=2,
    linewidth=2,
    label="user function graph"
)
plt.axhline(0,label='x-axis',color='black')
plt.axvline(0,label='y-axis',color='black')

plt.xlabel("x")
plt.ylabel("y")
plt.title("RK2 Method with Tolerance")

plt.grid(True)
plt.legend()

plt.show()
```

**QUESTION:**

Write a program to solve the ODE
dy/dx = x+y ; y(0)=1
corresponding exact solution:

           **y= 2e^x-x-1**

using the RK2 (Improved Euler) Method with (h=0.1) up to (x=0.5). Calculate for the exact solution;absolute error, percentage error, maximum absolute error, and maximum percentage error. Plot the RK2 approximation vs. exact solution and the absolute error vs. (x).

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x, y):
    return x + y
x0 = 0
y0 = 1
h = 0.1
x_final = 0.5
n = int((x_final - x0) / h)
x_values = [x0]
y_values = [y0]
x = x0
y = y0

for i in range(n):

    k1 = f(x, y)
    y_predict = y + h * k1
    k2 = f(x + h, y_predict)
    y_new = y + (h / 2) * (k1 + k2)
    x = x + h
    y = y_new
    x_values.append(x)
    y_values.append(y)

    print(
        "Iteration =", i + 1,
        ", x =", x,
        ", y =", y,
        ", k1 =", k1,
        ", k2 =", k2)

def exact_solution(x):
    return 2 * np.exp(x) - x - 1

errors = []
percentage_errors = []

print("\nError Analysis:")

for i in range(len(x_values)):
    exact_y = exact_solution(x_values[i])

    error = abs(exact_y - y_values[i])

    # Percentage error
    if exact_y != 0:
        percentage_error = (error / abs(exact_y)) * 100
    else:
        percentage_error = 0

    errors.append(error)
    percentage_errors.append(percentage_error)

    print(
        "x =", x_values[i],
        ", RK2 =", y_values[i],
        ", Exact =", exact_y,
        ", Error =", error,
        ", Percentage Error =", percentage_error, "%"
    )

# Maximum errors
max_error = max(errors)
max_percentage_error = max(percentage_errors)

print("\nMaximum Absolute Error =", max_error)
print("Maximum Percentage Error =", max_percentage_error, "%")

x_plot = np.linspace(x0, x_final, 300)

y_exact_plot = exact_solution(x_plot)

plt.figure(figsize=(8, 6))

plt.plot(
    x_plot,
    y_exact_plot,
    linewidth=2,
    label="Exact Solution",
    color='green')

plt.plot(
    x_values,
    y_values,
    marker="o",
    linewidth=1,
    label="RK2 Approximation",
    color='purple')
plt.xlabel("x")
plt.ylabel("y")
plt.title("RK2 Method vs Exact Solution")

plt.grid(True)
plt.legend()

plt.show()

plt.figure(figsize=(8, 6))

plt.plot(
    x_values,
    errors,
    marker="o",
    linestyle="-",
    linewidth=2,
    label="RK2 Absolute Error")

plt.xlabel("x")
plt.ylabel("Absolute Error")
plt.title("RK2 Error Analysis")

plt.grid(True)
plt.legend()

plt.show()
```
