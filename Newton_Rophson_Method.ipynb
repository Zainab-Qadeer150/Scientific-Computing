**Newton Rophson Method:**

"The Newton–Raphson method is an iterative numerical method used to find an approximate root of a nonlinear equation"

f(x)=0.

It starts with an initial guess x0
and uses the tangent line to the function to obtain successively better approximations to the root."

**Points to be Remember:**


*  It is the best known non-linear equation solver for finding approximation to the root of real value function f(x).
*   It require only one initial estimate to start the approximation.


*   It converges quadratically and provided the initial estimate is sufficiently close to the root x*, and the root is simple.
*   If intial estimate is not close to the x* then the method may diverge.


*   It is not necessarily globally convergent.


**Formula:**

The formula is:

    xn+1=xn- f(xn)/f'(xn)


where:

xn = current approximation

xn+1= next approximation

f(xn) = function value

f'(xn)= derivative of the function

Step 1: Choose the nonlinear function f(x).

```python
# QUESTION NO 01:
# Find the real root of the following equation using Newton Rophsan Method (correct upto 6dp):
# x^3-x-2

import numpy as np

def f(x):
    return x**3 - x - 2

def df(x):
    return 3*x**2 - 1

x = 1.5

tol = 1e-6
max_iter = 100

for i in range(max_iter):

    x_new = x - f(x)/df(x)

    if abs(x_new - x) < tol: #if the difference between the new value and old value is very small, it means we reached our convergence.
        break

    x = x_new
print("Root =", x_new)
print("Iterations =", i+1)
```

```python
# QUESTION NO 02:
# Write a Python program that uses the Newton-Raphson method to find the root of a user-defined nonlinear function,
# automatically calculates its derivative, checks convergence, displays the iterations,
# and plots the function with the initial estimate and approximate root.

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

#  Create the symbolic variable


x = sp.symbols('x')

function_input = input("Enter your function f(x): ")

#  Convert input into SymPy expression

f_sym = sp.sympify(function_input)


#  Automatically calculate derivative


df_sym = sp.diff(f_sym, x)


print("\nFunction:")
print("f(x) =", f_sym)

print("\nDerivative:")
print("f'(x) =", df_sym)

#  Convert SymPy expressions into
#    numerical functions


f = sp.lambdify(x, f_sym, modules="numpy")
df = sp.lambdify(x, df_sym, modules="numpy")

x0 = float(input("\nEnter initial estimate x0: "))


tol = 1e-6
max_iter = 100

# Check whether initial estimate
#    is already a root


if abs(f(x0)) < tol:

    print("\nInitial estimate is already a root.")
    print("Root =", x0)
    print("Iterations = 0")


else:

    x_current = x0

    # Store all approximations
    points = [x0]


    for i in range(max_iter):


        if abs(df(x_current)) < 1e-12:

            print("\nDerivative is zero or extremely small.")
            print("Newton-Raphson cannot continue.")

            break

        # Newton-Raphson formula


        x_new = x_current - f(x_current) / df(x_current)

        # Store new approximation

        points.append(x_new)


        print("Iteration", i + 1, ":", x_new)


        # Check convergence

        if abs(x_new - x_current) < tol:

            print("\nRoot =", x_new)
            print("Iterations =", i + 1)
            print("f(root) =", f(x_new))

            x_current = x_new

            break

        # Update current approximation

        x_current = x_new

    #  Plot the function


    root = x_current

    X = np.linspace(root - 2, root + 2, 400)

    Y = f(X)


    plt.figure(figsize=(8, 6))


    # Function
    plt.plot(
        X,
        Y,
        linewidth=2,
        label=f"f(x) = {f_sym}"
    )


    # x-axis
    plt.axhline(
        0,
        linewidth=1
    )


    # Root
    plt.scatter(
        root,
        f(root),
        s=100,
        label="Root"
    )


    # Initial estimate
    plt.scatter(
        x0,
        f(x0),
        s=100,
        label="Initial estimate"
    )

    plt.title("Newton-Raphson Method")
    plt.xlabel("x")
    plt.ylabel("f(x)")

    plt.grid(True)
    plt.legend()

    plt.show()
```

```python
# QUESTION NO 04:
# Use the Newton-Raphson method to find an approximate root of the nonlinear function
# sin(x)+x^2-cos(x)+e^(x^2)
# starting from the initial guess (x_0=1.7). Define the corresponding derivative (f'(x)), perform a maximum of 20 iterations
# stop the iteration when the difference between two consecutive approximations is less than (10^-7).
# Finally, plot the function, mark the initial guess and approximate root
# show the tangent line at the initial guess to illustrate the Newton-Raphson method graphically.

import numpy as np
import matplotlib.pyplot as plt
def f(x):
  return np.sin(x)+x**2-np.cos(x)+np.exp(x**2)
def df(x):
  return np.cos(x)+2*x+np.sin(x)+2*x*np.exp(x**2)
x= 1.7
tol = 1e-7
for i in range(20):
  x_new= x - f(x)/df(x)
  if abs(x_new-x)<tol:
    break
  x=x_new
root=x_new
X = np.linspace(0,8,300)
Y = f(X)

plt.plot(X,Y,
         color='green',
         linewidth=2,
         label='f(x)=x³-x-2')

plt.axhline(0,color='black')

plt.scatter(root,
            f(root),
            color='orange',
            s=100,
            label='Root')

plt.scatter(1.7,
            f(1.7),
            color='blue',
            s=100,
            label='Initial Guess')

m = df(1.7)

y_tangent = m*(X-1.7) + f(1.7)

plt.plot(X,
         y_tangent,
         '--',
         color='black',
         label='Tangent')

plt.title("Newton-Raphson Method")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.legend()

plt.show()
```

```python
# QUESTION NO 05:
# Develop a Python program that uses the Newton-Raphson method to find the approximate root of a user-defined nonlinear function.
# The program should allow the user to enter the function and an initial estimate, automatically calculate the derivative using SymPy, and convert both the function and derivative into numerical functions.
# The program should iteratively calculate the root, check for a zero or extremely small derivative, stop when the required tolerance is achieved, display the iteration results.
# plot the function along with the initial estimate, Newton-Raphson approximations, and the final root.

import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

# 1. Create symbolic variable
x = sp.symbols('x')

# 2. Take function from user
function = input("Enter the function: ")

# 3. Convert input into SymPy expression
f_symp = sp.sympify(function)

# 4. Automatically calculate derivative
derivative = sp.diff(f_symp, x)

print("\nFunction:")
print("f(x) =", f_symp)

print("Derivative:")
print("f'(x) =", derivative)

# 5. Convert SymPy expressions into numerical functions
g = sp.lambdify(x, f_symp, modules="numpy")
dg = sp.lambdify(x, derivative, modules="numpy")

# 6. Take initial estimate from user
x_0 = float(input("Enter your number: "))

# 7. Set tolerance and maximum iterations
tol = 1e-6
max_iter = 100

# Check whether initial estimate is already a root
if abs(g(x_0)) < tol:

    print("Initial estimate is already a root.")
    print("Root =", x_0)
    print("Iterations = 0")

    root = x_0

else:

    x_current = x_0

    # Store all approximations
    points = [x_0]

    for i in range(max_iter):

        # Check derivative before division
        if abs(dg(x_current)) < 1e-13:

            print("Derivative is zero or extremely small.")
            print("Newton-Raphson cannot continue.")
            break

        # Newton-Raphson formula
        x_new = x_current - g(x_current) / dg(x_current)

        # Store approximation
        points.append(x_new)

        # Display iteration
        print("Iteration", i + 1, ":", x_new)

        # Check convergence
        if abs(x_new - x_current) < tol:

            root = x_new

            print("\nRoot =", root)
            print("Iterations =", i + 1)
            print("f(root) =", g(root))

            break

        # Update current approximation
        x_current = x_new

    else:

        root = x_current

        print("\nMaximum number of iterations reached.")
        print("Approximate root =", root)

# Graph

X = np.linspace(root - 2, root + 2, 400)

Y = g(X)

plt.figure(figsize=(8, 6))

# Function
plt.plot(
    X,
    Y,
    linewidth=2,
    label=f"f(x) = {f_symp}",
    color="green"
)

# x-axis
plt.axhline(
    0,
    linewidth=1,
    color="pink"
)

# Root
plt.scatter(
    root,
    g(root),
    s=100,
    label="Root",
    color="yellow"
)

# Initial estimate
plt.scatter(
    x_0,
    g(x_0),
    s=100,
    label="Initial estimate",
    color="black"
)

# Newton-Raphson approximations
plt.scatter(
    points,
    [g(p) for p in points],
    s=30,
    label="Newton-Raphson approximations"
)

plt.title("Newton-Raphson Method")
plt.xlabel("x")
plt.ylabel("f(x)")

plt.grid(True)
plt.legend()

plt.show()
```
