**Bisection Method for Numerical Root Finding:**


"The Bisection Method is a reliable bracketing-based numerical technique for finding a root of a nonlinear equation f(x)=0. The method repeatedly divides an interval into two halves and retains the subinterval in which a sign change occurs."

 **Benefits and Drawbacks:**

 1. The bisection method is the simplest method for finding a zero of continuous  real-valued function which is repeatedly bisects an interval, then selects a sub-interval in which zero is known to occur for further processing.

 2. It is known to be the most robust method and also globally convergent.

 3. One of the main drawback is that  is relatively slow.




**Algorithm:**
1. Choose a and b.
2. Check f(a)f(b) < 0.
3. Calculate midpoint c.
   by using this formula:
   c= a+b/2
4. Evaluate f(c).
5. Select the interval containing the sign change.
6. Repeat until the error is below the tolerance.
7. Report the root and number of iterations.

```python
# Question no 01:   Find a real root of the following equation that correct up to (6 decimal places) using bisection method.
# x^3-x-2
def f(x):
  return x**3-x-2
a=1
b=2
tol= 1e-6
#tol = 0.000001

# It shows 10⁻⁶ accuracy,we need.
max_iter = 100
for i in range(max_iter):
  c= (a+b)/2
  if abs(f(c))<tol:
    break
  if f(a)*f(c)<0:
    b=c
  else:
    a=c
print("root= ",c)
print("iterations= ",i+1)
```

```python
# Question no 02:
# Find a real root of the following equation that correct up to 6 decimal places using bisection method and show the root graphically?
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x**3 - x - 2

x = np.linspace(1, 2, 200)
y = f(x)# for all 200 x values

root = 1.52138 # we store in variable to see where the root crosses the x-axis

plt.plot(x, y,
         color="blue",
         linewidth=2,
         label="f(x)=x³-x-2")

plt.scatter(root, f(root) # f(root) shows that the root are almost zero
            ,color="red",
            s=80,
            label="Root")

plt.axhline(0, color="black") #Because the root is where

#f(x)=0

# Without this line, it is harder to see where the function crosses the x-axis.

plt.title("Bisection Method")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.legend()

plt.show()
```

```python
# Question no 5:
# Develop a user-input-based Python program to numerically determine the root
# of a continuous nonlinear equation using the Bisection Method.The program should
# an initial interval, and a tolerance,and visualize the approximate root of the function.
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Create symbolic variable

x = sp.symbols('x')

function_input = input("Enter your function f(x): ")

# Convert user input into SymPy expression
f_sym = sp.sympify(function_input)

print("\nf(x) =", f_sym)

#  Convert SymPy expression
#    into numerical function
f = sp.lambdify(x, f_sym, modules="numpy")

a = float(input("Enter lower bound a: "))
b = float(input("Enter upper bound b: "))

tol = float(input("Enter tolerance: "))

max_iter = 100


if f(a) * f(b) > 0:

    print("\nNo sign change between a and b.")
    print("Bisection Method cannot be applied.")

else:

    # Store bisection approximations
    points = []

    for i in range(max_iter):

        # Calculate midpoint
        c = (a + b) / 2

        # Store midpoint
        points.append(c)

        print(
            "Iteration", i + 1,
            ": a =", a,
            ", b =", b,
            ", c =", c,
            ", f(c) =", f(c)
        )

        # To Check convergence
        if abs(f(c)) < tol:

            root = c

            print("\n We achieved the Convergence.")
            print("Root =", root)
            print("Iterations =", i + 1)

            break

        if f(a) * f(c) < 0:

            b = c

        else:

            a = c

    else:

        root = c

        print("\nMaximum number of iterations reached.")
        print("Approximate root =", root)

    # 7. Plot function

    X = np.linspace(a - 1, b + 1, 500) # to make a graph smooth and x,y points does'nt compact with each other that'why we used.
    Y = f(X)

    plt.figure(figsize=(8, 6))

    # Function
    plt.plot(
        X,
        Y,
        linewidth=2,
        label=f"f(x) = {f_sym}",
        color='green'
    )

    # x-axis
    plt.axhline(
        0,
        linewidth=1,
        label="x-axis",
        color='red'
    )

    # Root
    plt.scatter(
        root,
        f(root),
        s=80,
        label="Root",
        color='blue'
    )

    # Bisection approximations
    plt.scatter(
        points,
        [f(p) for p in points],
        s=25,
        label="Bisection approximations",
        color='purple'
    )

    plt.xlabel("x")
    plt.ylabel("f(x)")
    plt.title("Bisection Method")

    plt.grid(True)
    plt.legend()

    plt.show()
```
