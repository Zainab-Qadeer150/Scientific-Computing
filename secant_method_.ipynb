**Secant Method:**

​Secant method is an extrapolation method,that require two intial guess to start the next approximation to the zero.It has golden ration i.e 1.62,it is superlinear but not as good as quadratic.

**  FORMULA:**

xn+1 = xn - f(xn) * (xn - xn-1) / (f(xn) - f(xn-1))

**Algorithm**

Step 1: Given a nonlinear equation

f(x)=0

Step 2: Choose two initial approximations x0
and x
1

Step 3: Calculate

f(x0),f(x1)

Step 4: Calculate the next approximation using

xn+1 = xn - f(xn) * (xn - xn-1) / (f(xn) - f(xn-1))
Step 5: Check the stopping condition, for example:

∣xn+1−xn∣<ε

where ε is the required tolerance.

Step 6: If the stopping condition is not satisfied, set

xn−1=xn

and

xn= xn+1
	​


Then repeat Step 4.

Step 7: When the stopping condition is satisfied, take xn+1
 as the approximate root.

**Secant Method via Newton Raphson Method**

| Secant Method                       | Newton-Raphson                              |
| ----------------------------------- | ------------------------------------------- |
| Uses two initial guesses            | Uses one initial guess                      |
| Does **not** require (f'(x))        | Requires (f'(x))                            |
| Uses (f(x_n)) and (f(x_{n-1}))      | Uses (f'(x_n))                              |
| Generally faster than bisection     | Generally faster than secant                |
| Easier when derivative is difficult | Very efficient when derivative is available |


```python
# QUESTION NO 01:
# Use the Secant Method to find the root of the nonlinear equation i.e x^3-2x+4
# Implement the Secant Method in Python and display the approximate root and the number of iterations required for convergence.

def f(x):
  return x**3-2*x+4
x0=1
x1=2
tol= 1e-6
max_iter=100
for i in range(max_iter):
  x2=x1- f(x1)* (x1-x0)/(f(x1)-f(x0))
  print("iterations = ", i+1,": ", x2)
  if abs(x2-x1)<tol:
    break
  x0=x1
  x1=x2
print("root ", x2)
print("iteration= " , i+1)







```

```python
# QUESTION NO 02:
# Use the Secant Method to find the root of (f(x)=x^3-2x+4=0) with initial guesses (x_0=1) and (x_1=2).
# Use a tolerance of (10^{-6}) and a maximum of 100 iterations.
# Implement the method in Python, display the iteration values and final root, and plot the function, initial guesses, Secant approximations, and computed root.
import numpy as np
import matplotlib.pyplot as plt
def f(x):
  return x**3-2*x+4
x0=1
x1=2
tol= 1e-6
max_iter=100
# store all secant approximation
points = [x0, x1]
for i in range(max_iter):
  denominatior = f(x1)-f(x0)
  if denominatior == 0:
    print("Denominator is zero")
    break
  x2=x1- f(x1)* (x1-x0)/denominatior
  points.append(x2)
  print("iteration = ", i+1, ": ", x2)
  if abs(x2-x1)<tol:
    break
  x0=x1
  x1=x2
  root=x2
print("\nroot ", x2)
print("iteration= " , i+1)
X=np.linspace(-1,60,200)
Y=f(X)
plt.figure(figsize=(12, 7))
plt.axhline(0,linewidth=1,color='red',label="x-axis")
plt.plot(X,Y,linewidth=2,color='green',label="x**3-2*x+4")
plt.scatter(x0,f(x0),label="intial guess",s=300,color='black')
plt.scatter(root,f(root),label="root",s=70,color='purple')
plt.scatter(points,[f(p) for p in points],label="secant approximation",s=15,color='blue')
plt.title("Secant Method")
plt.xlabel("x")
plt.ylabel("f(x)")

plt.grid(True)
plt.legend()

plt.show()
```

```python
# QUESTION NO 03:
# f(x)=e^x+x^2-4=0
# using initial guesses (x_0=1) and (x_1=3), a tolerance of (10^{-6}), and a maximum of 100 iterations.
# Implement the method in Python and plot the function, initial guesses, Secant approximations, root,
# and Secant lines for each iteration.

import numpy as np
import matplotlib.pyplot as plt
def f(x):
  return np.exp(x)+x**2-4
x0=1
x1=3
tol = 1e-6
max_iter = 100
points=[x0,x1]
secant_line=[]
for i in range(max_iter):
  denominator= f(x1)-f(x0)
  if denominator==0:
    print("Cannot continue: denominator is zero.")
    break
   # Calculate slope of secant line
  m = (f(x1) - f(x0)) / (x1 - x0)
  # Store information for this secant line
  secant_line.append((x0, x1, m))
  x2 = x1 - f(x1) * (x1 - x0) / denominator
  points.append(x2)
  if abs(x1-x0)<tol:
    break
  x0=x1
  x1=x2
root=x2
print("\nroot", x2)
print("iteration= ", i+1)
X=np.linspace(-1,60,200)
Y=f(X)
plt.figure(figsize=(12,7))
plt.axhline(0,linewidth=1,color='red',label="x-axis")
plt.plot(X,Y,linewidth=2,color='green',label="x**3-2*x+4")
plt.scatter(x0,f(x0),label="intial guess",s=300,color='gray')
plt.scatter(root,f(root),label="root",s=70,color='purple')
plt.scatter(points,[f(p) for p in points],label="secant approximation",s=15,color='blue')
for j,(a,b,m) in enumerate(secant_line):
  Y_secant=f(a) + m * (X - a)
  plt.plot(X,Y_secant,linestyle="--",linewidth=1,label=f"Secant {j+1}",color='black')
plt.title("Secant Method with Iteration Lines")
plt.xlabel("x")
plt.ylabel("f(x)")

plt.grid(True)
plt.legend()

plt.show()
```

```python
# QUESTION NO 04:
# Develop a Python program to implement the Secant Method for finding the root of a user-defined nonlinear equation.
# Allow the user to enter the function and two initial guesses, use a tolerance of (10^{-6}) and a maximum of 100 iterations.
# Display the iteration results and final root, and plot the function, initial guesses, Secant approximations, and computed root.

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt


# 1. Create symbolic variable
x = sp.symbols("x")


# 2. Take function from user
function = input("Enter your function: ")

f_sym = sp.sympify(function)

print("\nFunction:")
print("f(x) =", f_sym)


# 3. Convert symbolic function
#    into numerical function
f = sp.lambdify(x, f_sym, modules="numpy")


# 4. Take two initial guesses
x0 = float(input("Enter your first initial guess: "))
x1 = float(input("Enter your second initial guess: "))


# 5. Settings
tol = 1e-6
max_iter = 100


# 6. Store approximations
points = [x0, x1]


# 7. Secant Method
for i in range(max_iter):

    denominator = f(x1) - f(x0)

    if abs(denominator) < 1e-12:
        print("\nDenominator is extremely small or zero.")
        break

    # Secant formula
    x2 = x1 - f(x1) * (x1 - x0) / denominator

    # Store new approximation
    points.append(x2)

    print("Iteration", i + 1, ":", x2)

    # Check convergence
    if abs(x2 - x1) < tol:
        print("\nConvergence is achieved.")
        break

    # Move points forward
    x0 = x1
    x1 = x2


# 8. Final root
root = x2

print("\nRoot =", root)
print("Iterations =", i + 1)


# 9. Create graph
X = np.linspace(root - 2, root + 2, 500)
Y = f(X)


# 10. Plot
plt.figure(figsize=(12, 7))

plt.axhline(
    0,
    linewidth=1,
    color="red",
    label="x-axis"
)

plt.plot(
    X,
    Y,
    linewidth=2,
    color="green",
    label=f"f(x) = {f_sym}"
)


# Initial guesses
plt.scatter(
    points[0],
    f(points[0]),
    label="Initial guess x0",
    s=100,
    color="gray"
)

plt.scatter(
    points[1],
    f(points[1]),
    label="Initial guess x1",
    s=100,
    color="orange"
)


# Root
plt.scatter(
    root,
    f(root),
    label="Root",
    s=100,
    color="purple"
)


# All Secant approximations
plt.scatter(
    points,
    [f(p) for p in points],
    label="Secant approximations",
    s=10,
    color="blue"
)


plt.title("Secant Method")
plt.xlabel("x")
plt.ylabel("f(x)")

plt.grid(True)
plt.legend()

plt.show()
```
